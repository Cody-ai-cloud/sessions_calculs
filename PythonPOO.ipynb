```python
class Moteur:
    def __init__(self, nom, poussee_kN):
        self.nom = nom
        self.poussee_kN = poussee_kN
        self.allume = False

    def demarrer(self):
        self.allume = True
        return f"{self.nom} demarre"

    def poussee_a(self, altitude_m):
        facteur = max(0.0, 1 - altitude_m / 20000)
        return self.poussee_kN * facteur

m = Moteur("CFM56", 120)
print(m.demarrer())        # CFM56 demarre
print(m.allume)            # True
print(m.poussee_a(10000))  # 60.0
```

```python
class Reservoir:
    def __init__(self, capacite_l):
        self.capacite_l = capacite_l
        self._niveau_l = 0.0          # interne

    @property
    def niveau_pourcent(self):
        return 100 * self._niveau_l / self.capacite_l

    def remplir(self, litres):
        if litres < 0:
            raise ValueError("litres doit etre positif")
        self._niveau_l = min(self.capacite_l, self._niveau_l + litres)

r = Reservoir(2000)
r.remplir(500)
print(r.niveau_pourcent)   # 25.0  <- pas de parentheses !
# Revoir les liaisons en tre fonction et attributs
```

```python
class Aeronef:
    def __init__(self, immat, masse_kg):
        self.immat = immat
        self.masse_kg = masse_kg

    def decoller(self):
        return f"{self.immat} decolle"

    def decrire(self):
        return f"{self.immat} ({self.masse_kg} kg)"


class Planeur(Aeronef):
    def __init__(self, immat, masse_kg, finesse):
        super().__init__(immat, masse_kg)   # on appelle le parent
        self.finesse = finesse

    def decoller(self):                     # on redefinit (override)
        return f"{self.immat} est treuille, puis decolle"


p = Planeur("F-CABC", 450, 40)
print(p.decoller())   # version du Planeur
print(p.decrire())    # version heritee de l'Aeronef, gratuite
```

```python
flotte = [
    Aeronef("F-GKXA", 70000),
    Planeur("F-CABC", 450, 40),
]

for a in flotte:
    print(a.decoller())   # Python choisit la bonne version tout seul 
```

```python
class Avion:
    def __init__(self, immat, moteurs):
        self.immat = immat
        self.moteurs = moteurs          # une liste d'objets Moteur

    def poussee_totale(self):
        return sum(m.poussee_kN for m in self.moteurs)

a320 = Avion("F-GKXA", [Moteur("CFM56", 120), Moteur("CFM56", 120)])
print(a320.poussee_totale())   # 240
```
